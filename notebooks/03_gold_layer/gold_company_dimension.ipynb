```python
from IPython.core.magic import register_cell_magic

@register_cell_magic
def dsql(line, cell):
    df = conn.sql(cell).df()
    var_name = line.strip()
    if var_name:
        globals()[var_name] = df
    return df
```

```python
conn.sql(""" 
         CREATE OR REPLACE TABLE gold.dim_filing AS 
         SELECT DISTINCT
         adsh AS accession_number,
         cik AS company_id,
         name AS company_name,
         s.sic AS industry_code,
         l.industry_name AS industry_name,
         l.industry_group AS industry_group,
         form AS filing_type,
         period AS period_end,
         fy AS fiscal_year,
         fp AS fiscal_period,
         fye AS fiscal_year_end,
         filed AS filing_date,
         prevrpt AS is_previous_report,
         FROM clean.sub s
         JOIN main.sic_lookup l 
         ON s.sic = l.sic
         """)
```
